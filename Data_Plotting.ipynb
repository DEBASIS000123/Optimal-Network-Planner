```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
#All the Data in SupplementaryMaterial
data=pd.read_csv("D:/SupplementaryMaterial.csv");
data
```

```python
#All the columns in Data file
data.columns
```

```python
features = data[['Longitude', 'Latitude', 'Serving RSSI_4G']]
features.rename(columns={'Serving RSSI_4G': 'RSSI'}, inplace=True)  # Rename column to 'RSSI'

# Heatmap of features
features.replace('\t', pd.NA, inplace=True)
features.replace(' ', pd.NA, inplace=True)
features = features.apply(pd.to_numeric, errors='coerce')

p1 = sns.heatmap(features)
plt.savefig('D:/Plots/heatmap_features.jpg', format='jpg', dpi=300, bbox_inches='tight')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Prepare data for 3D scatter plot
x = data['Longitude'].replace(['\t', ' '], pd.NA).apply(pd.to_numeric, errors='coerce')  # Longitude on X-axis
y = data['Latitude'].replace(['\t', ' '], pd.NA).apply(pd.to_numeric, errors='coerce')   # Latitude on Y-axis
z = data['Serving RSSI_4G'].replace(['\t', ' '], pd.NA).apply(pd.to_numeric, errors='coerce')  # Serving RSRP_4G on Z-axis
colors = z  # Use Serving RSRP_4G values for color mapping

# Create a 3D scatter plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')  # Create a 3D axis

# Scatter plot
scatter = ax.scatter(x, y, z, c=colors, cmap='coolwarm', s=30, edgecolor='k', alpha=0.8)

# Add colorbar
cbar = fig.colorbar(scatter, ax=ax, shrink=0.6, aspect=20, pad=0.15)
cbar.set_label("RSSI", fontsize=20, weight='bold')
cbar.ax.tick_params(labelsize=15)  # Increase colorbar tick labels size

# Customize axes
ax.set_xlabel("Longitude", fontsize=20, labelpad=15, weight='bold')
ax.set_ylabel("Latitude", fontsize=20, labelpad=15, weight='bold')
ax.set_zlabel("RSSI", fontsize=20, labelpad=18, weight='bold')

# Increase tick label font size
ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=12)
ax.tick_params(axis='z', labelsize=12)

# Add title
ax.set_title("3D Scatter Plot: Longitude vs Latitude vs RSSI", fontsize=25, weight='bold', pad=20)

# Adjust view angle for better visualization
ax.view_init(elev=30, azim=135)

# Save the 3D plot as a .jpg file
plt.savefig("D:/Plots/3d_scatter_plot.jpg", format='jpg', dpi=600, bbox_inches='tight')

# Show the plot
plt.show()
```
